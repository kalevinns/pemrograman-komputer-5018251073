1. Perbedaan int dan float: int (integer) digunakan untuk menyimpan bilangan bulat (contoh: 10, -5), sedangkan float digunakan untuk menyimpan bilangan desimal atau pecahan (contoh: 3.22 , -0.4).

2. Alasan konversi input() menggunakan float(): Secara bawaan semua data yang ditangkap oleh fungsi input() berjenis teks (string). Agar teks berupa angka tersebut bisa dihitung menggunakan operasi matematika, datanya harus diubah dulu menjadi tipe angka numerik, yaitu float().

3. Perbedaan = dan ==: Operator = berfungsi untuk memasukkan/menyimpan suatu nilai ke dalam variabel. Sebaliknya, operator == berfungsi sebagai perbandingan untuk mengecek apakah nilai di sisi kiri sama persis dengan nilai di sisi kanan.

4. Fungsi and dan or: and akan menghasilkan nilai True hanya jika semua kondisi bernilai benar. Sedangkan or akan menghasilkan True jika salah satu saja dari kondisi tersebut bernilai benar
